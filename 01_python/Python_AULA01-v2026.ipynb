## Variáveis

É a identificação de uma região de memória por um nome. Uma variável armazena temporariamente valores para serem usados ao longo de um programa.

### Observação na criação de variáveis

Existem algumas palavras que são reservadas no Python e por isso não podem ser utilizadas como nome de variáveis.
Para listá-las, execute o código abaixo:

````
import keyword
print(keyword.kwlist) # List all Python Keywords
````

### Boas práticas na criação de variáveis


**Nomes de variáveis:** Os nomes das variáveis em Python podem conter letras (maiúsculas e minúsculas), dígitos e underscores (_). No entanto, o nome de uma variável não pode começar com um número. Exemplos válidos de nomes de variáveis são idade, nome_completo, total_1.

**Sensibilidade a maiúsculas e minúsculas:** Python é sensível a maiúsculas e minúsculas, o que significa que variavel e Variavel são considerados diferentes.

**Palavras-chave reservadas:** Você não pode usar palavras reservadas do Python como nomes de variáveis. Por exemplo, você não pode nomear uma variável if, for, while, import, def, etc.

**Convenções de nomenclatura:** Embora Python não exija, é uma boa prática seguir convenções de nomenclatura para tornar seu código mais legível. Recomenda-se usar nomes de variáveis em minúsculas, separando palavras com underscores para melhor legibilidade (snake_case). Por exemplo, nome_completo, idade_usuario.

**Inicialização de variáveis:** Você pode inicializar uma variável atribuindo um valor a ela. Por exemplo, idade = 25, nome = "João".

### Atribuição de Variável

A atribuição de variável é o ato de associar um valor a um nome de variável em um programa de computador. Em muitas linguagens de programação, incluindo Python, a atribuição é feita usando o operador de atribuição, que é geralmente o sinal de igual (=).

````
intvar = 10 # Variável de valor inteiro
floatvar = 2.57 # Variável de valor float - ponto flutuante
strvar = "Python Language" # Variável string - texto
print(intvar)
print(floatvar)
print(strvar)
````

#### Atribuição Múltiplas

Em Python é possível atribuir múltiplos valores a múltiplas variáveis em uma única linha de código.

````
# Usando vírgulas para separar os valores, atribuímos 10 à intvar, 2.57 à floatvar e "Python Language" a strvar
intvar , floatvar , strvar = 10, 2.57, "Python Language"

print(intvar)
print(floatvar)
print(strvar)
````

````
p1 = p2 = p3 = p4 = 44 # Todas as variáveis apontando para o mesmo valor
print(p1, p2, p3, p4)
````

## Tipagem de Dados em Python

A tipagem em Python é **dinâmica e inferida**.

Você não precisa declarar explicitamente o tipo de uma variável ao criá-la. O tipo é inferido a partir do valor atribuído à variável. Isso torna Python flexível e fácil de usar, mas também pode resultar em erros de tipo que só são detectados durante a execução do programa.

````
x = 0 # x é vinculado a um objeto inteiro
print('Primeira atribuição >>>>>>>>', type(x))
x = "Olá" # Agora ele é uma string
print('Segunda atribuição >>>>>>>>', type(x))
x = [1, 2, 3] # E agora ele é uma lista
print('Terceira atribuição >>>>>>>>', type(x))
````


## Strings

De início vamos aprender uma estrutura de dados: **Strings**.

**Strings** é um conjunto ordenado de caracteres, usado para armazenar e representar informações textuais.

De modo geral, as strings são muito fáceis de usar. Talvez o mais complicado a respeito delas seja que existem muitas maneiras de escrevê-las em seu código:

````
s1 = 'spa"m' # apostrofos
s2 = "spa'm" # aspas
s3 = '''...spam...'''; s4 = """...spam...""" # apostrofos ou aspas triplas
s5 = "s\tp\naom" # sequencia de escape
s6 = r"C:\new\test.spm" # strings brutas
print(s1)
print(s2)
print(s3)
print(s4)
print(s5)
print(s6)
````

**s1** é uma string definida usando apostrofos ('). A string contém aspas duplas dentro dela.

**s2** é uma string definida usando aspas ("). Neste caso, a string contém um apóstrofo dentro dela.

**s3 e s4** são strings definidas usando aspas triplas (' ' ' ou " " "). Elas são usadas quando a string precisa abranger múltiplas linhas ou contém muitas aspas ou apóstrofos. Elas podem ser usadas com aspas ou apóstrofos triplos, e a escolha entre os dois depende das preferências do programador.

**s5** é uma string que contém sequências de escape, indicadas por barras invertidas (\\). Por exemplo, \t representa uma tabulação e \n representa uma nova linha. Essas sequências são interpretadas de maneira especial pelo interpretador Python quando a string é processada.

**s6** é uma string bruta, definida com o prefixo r. Isso significa que as sequências de escape dentro da string não são interpretadas, tornando-a útil para lidar com caminhos de arquivos no Windows, por exemplo, onde barras invertidas são comumente usadas. Neste caso, as barras invertidas são tratadas literalmente e não como uma sequência de escape.

### Strings com apóstrofos e aspas são iguais.

````
'palavra', "palavra"
````

````
'knight"s', "knight's"
````

### Apóstrofos ou Aspas triplas definem strings de bloco de várias linhas

````
# Cria a variável texto do tipo string contendo várias linhas
texto_aspas="""
Rodou o mundo, indo sempre ao leste.
Fez questão de nunca errar a latitude no seu GPS.
Compensava meticulosamente a rota tomando referências precisas da sua última posição.
Vigiava as estrelas e a pequena bússola, Mas nunca atingiu a borda do mundo
Tolo. Voltou paro mesmo ponto. A terra é redonda!
"""
print(texto_aspas)

texto_apostrofos='''
Rodou o mundo, indo sempre ao leste.
Fez questão de nunca errar a latitude no seu GPS.
Compensava meticulosamente a rota tomando referências precisas da sua última posição.
Vigiava as estrelas e a pequena bússola, Mas nunca atingiu a borda do mundo
Tolo. Voltou paro mesmo ponto. A terra é redonda!
'''
print(texto_apostrofos)
````


## Escrevendo e Capturando dados do usuário

### print - escrevendo dados

*print* é o comando geral para imprimir dados, sentenças, o que for para o usuário. a sintaxe mais simples é invocar passando o que precisa ser escrito como seu argumento

Vamos a alguns exemplos utilizando o print.

````
print("Olá, mundo!") # Print de um valor diretamente

nome = "Alice"
print(nome) # Print de uma variável

a = 5
b = 3
print(a + b) # Print de uma expressão

saudacao = "Olá"
nome = "Alice"
idade = 30
print(saudacao + ", meu nome é " + nome + " e eu tenho " + str(idade) + " anos.") # Print de uma combinação de valores, variáveis e expressões

nome = "Bob"
idade = 25
print(f"Meu nome é {nome} e eu tenho {idade} anos.") # Print formatado (Python 3.6+)
# Em Python, o f antes de uma string dentro de um comando print é chamado de f-string (format string).
# As f-strings são uma maneira conveniente e legível de formatar strings que foram introduzidas na versão 3.6 do Python.

nome = "Carol"
idade = 40
print("Meu nome é {} e eu tenho {} anos.".format(nome, idade)) # Print usando a função format
````

````
nome = "Pedro"

print(nome)
````

## Indentação e Blocos

A **Indentação** refere-se aos espaços no início de uma linha de código. É muito importante porque o Python usa
recuo para indicar um bloco de código. Se o recuo não estiver correto, terminaremos com Erro de **IndentationError**

````
p = 10
if p == 10:
 print ('P is equal to 10') # indentação correta
````

````
# se a indentação for ignorado, encontraremos IndentationError: expected an indented block
p = 10
if p == 10:
print ('P is equal to 10')
````

````
for i in range(0,5):
 print(i) # indentação correta
````

````
# se a indentação for ignorado, encontraremos IndentationError: expected an indented block
for i in range(0,5):
print(i)
````

````
for i in range(0,5): print(i) # indentação correta, mas menos legível
````

````
j=20
for i in range(0,5):
 print(i) # dentro da repetição for
print(j) # fora da repetição for
````

## Função dir() e dunders "__doc __"

A função **dir()** em Python é uma função integrada que retorna uma lista de todos os atributos e métodos disponíveis para um determinado objeto. Esses atributos incluem métodos especiais (também conhecidos como "métodos mágicos" ou "dunders") e métodos comuns.

Quando você chama **dir()** sem argumentos, ela retorna uma lista de nomes no escopo local atual. No entanto, quando você chama **dir()** com um argumento (um objeto), ela retorna uma lista de todos os atributos e métodos disponíveis para aquele objeto.

````
# Chamando dir() sem argumento
dir()
````

````
# Por exemplo, se você chamar dir(str), você verá todos os métodos disponíveis para string:

dir(str)
````

O __doc__ é um atributo especial que contém a documentação associada a um objeto, como uma função, classe ou módulo.

Este atributo armazena uma string de documentação, também conhecida como docstring, que descreve o propósito e o comportamento do objeto. A docstring é usada para fornecer informações sobre como usar o objeto, incluindo detalhes sobre seus parâmetros, retorno e funcionalidade em geral.

Quando você acessa:
```
#  len.__doc__
```
está acessando a docstring associada à função len().

````
print(len.__doc__)
````

Essa docstring descreve sucintamente o que a função len() faz. Ela informa que a função len() retorna o número de itens em um contêiner (container), que é um termo genérico para qualquer objeto que possa conter outros objetos, como listas, tuplas, strings, dicionários, etc.

Portanto,

```
# len.__doc__
```

 fornece uma descrição útil e concisa da função len(), ajudando os programadores a entenderem rapidamente seu propósito e como usá-la corretamente.
