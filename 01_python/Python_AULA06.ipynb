## O PyPI
Python Package Index, é o índice geral de pacotes do Python. Ali você irá encontrar TODOS os pacotes devidamente registrados e disponíveis publicamente para o Python. Para usar o PyPI, TODO editor Python tem um modulo ou uma interface para buscar e instalar os pacotes.

O Python tem uma interface de linha de comando própria para isso.

Como usar:
1. Procure pelo pacote ou alguma descrição.
2. Selecione o que estiver mais próximo do que busca
3. Não precisa baixar o arquivo. Basta ir no editor Python, ou na linha de comando e ativar o PyPI. Eles irão baixar automaticamente para você

Uma vez baixado, basta seguir a documentação do seu "criador" e pronto! Sair usando.

## Formas de importar um módulo
O comando para importar uma biblioteca é o *import*

Exitem 3 sintaxes para o import
### Importar a biblioteca somente
   Nesse caso, para usar um objeto desse modulo, você vai ter que acessar pelo nomeo do módulo.*nome_do_objeto*. No exemplo abaixo, estou invocando a função "func()" do módulo "modulo"
   ```
   # Aqui importo o módulo
   import modulo

   ...
   # Aqui eu invoco a função func() do modulo
   modulo.func()
   ```
### Importar a biblioteca mas atribuindo um apelido
   Nessa forma, além de importar, podemos dar um apelido. Imagine que você precise importar um módulo com um nome_excessivamente_longo. Será um horror se precisar usar esse módulo muitas vezes.
   ```
   import modulo_de_nome_longo as modulo
   ...
   modulo.func()
   ```

   A cláusula **as** dá um novo nome ao módulo.

   Dar apelido também serve para resolver conflitos de objetos; Imagine que o módulo X tenha uma função de nome func(); e a função Y, também tenha uma outra função de mesmo nome.  Se invocarmos somente * func()* de qual módulo ele vai puxar?

   Em programação nunca "pague pra ver"
   ```
   import modulo1 as m1
   import modulo2 as m2
   ..
   
   m1.func()
   m2.func()
   ```

### Importar uma operação somente da biblioteca
   Não é tão frequente, mas se o programador quiser somente buscar um mísero objeto de uma biblioteca. Nada o impede, por exemplo:
   ```
   from math import pi

   print(pi)
   ```

   Ele importará da biblioteca de funções e constantes matemáticas somente o valor de PI. Esse último voce pode rodar...

Quando usar uma delas, nunca se sabe. Vai de estilo para estilo de programação. Mas a regra é simples:
