## O que é IA Generativa

IA Generativa é o nome dado a sistemas de Inteligência Artificial capazes de gerar conteúdo novo — texto, código, imagens e até áudio — a partir de um comando em linguagem natural, chamado de **prompt**. As ferramentas mais conhecidas desse tipo são modelos de linguagem (LLMs, do inglês *Large Language Models*), como o ChatGPT (OpenAI), o Claude (Anthropic) e o Gemini (Google).

Diferente de uma calculadora ou de uma função do Pandas, esses modelos não seguem regras fixas: eles foram treinados com uma quantidade enorme de texto (incluindo código Python, documentação de bibliotecas e discussões técnicas) e aprenderam a **prever qual seria a próxima palavra mais provável** dado um contexto. Na prática, isso significa que você pode conversar com eles em português, pedir um gráfico, um trecho de código Pandas ou até uma explicação de um erro — e receber uma resposta coerente e, na maioria das vezes, correta.

## Por que usar IA Generativa na Análise de Dados

Ao longo deste curso, você aprendeu a escrever código Pandas, Seaborn, Matplotlib e Plotly do zero. Isso é essencial — sem esse conhecimento, você não teria como avaliar se o que a IA gera faz sentido. Mas, uma vez que você domina os fundamentos, a IA generativa pode acelerar bastante o seu dia a dia como Cientista de Dados:

- **Explorar hipóteses mais rápido** — perguntar qual tipo de gráfico melhor representa uma comparação, distribuição ou correlação
- **Gerar código boilerplate** — evitar reescrever do zero um `groupby()` ou uma configuração de `Figure` que você já viu antes
- **Explicar e corrigir erros** — colar uma mensagem de erro e entender rapidamente o que aconteceu
- **Documentar e resumir** — transformar uma análise técnica em um resumo executivo para quem não é técnico
- **Revisar seu próprio código** — pedir sugestões de simplificação ou boas práticas

> 📝 **Nota:** a IA generativa é uma ferramenta de **produtividade**, não um substituto para o seu raciocínio analítico. Ela escreve o código; você continua sendo responsável por validar se o resultado é correto e faz sentido para o problema.

## Como Escrever Bons Prompts para Análise de Dados

A qualidade da resposta da IA depende diretamente da qualidade do seu prompt. Um bom prompt para análise de dados costuma ter quatro elementos:

1. **Contexto** — qual é o dataset, quais colunas existem e o que elas representam
2. **Objetivo** — o que você quer descobrir ou construir
3. **Formato de saída** — você quer só uma sugestão, um trecho de código, uma explicação?
4. **Restrições** — qual biblioteca usar, qual estilo de código seguir, limitações do ambiente (ex: Google Colab)

Um template simples que funciona bem:

> "Tenho um DataFrame Pandas chamado `pokemons` com as colunas [liste as colunas relevantes]. Quero [objetivo]. Gere um código em Python usando [biblioteca] que faça isso, com comentários curtos explicando cada etapa."

> 💡 **Dica:** quanto mais específico for o seu prompt sobre as colunas e o objetivo, menor a chance da IA "alucinar" nomes de colunas que não existem no seu dataset.

## Preparando o Ambiente

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# carregando o mesmo dataset de pokemons utilizado nas aulas anteriores
pokemons = pd.read_csv('https://github.com/labeduc/datasets/blob/main/pokemons/all.csv?raw=true')

pokemons.head(3)
````

## Exemplo 1: Pedindo uma Sugestão de Gráfico

Vamos simular uma situação real: você quer comparar a média de ataque (`Att`) entre Pokémons lendários e não lendários, mas não tem certeza de qual gráfico usar. Em vez de abrir a documentação do Seaborn, você pode perguntar diretamente para a IA.

> **Prompt:** "Tenho um DataFrame de Pokémons com as colunas `Att` (ataque) e `Legendary` (1.0 para lendário, 0.0 para normal). Quero comparar a média de ataque entre os dois grupos. Que tipo de gráfico você recomenda e por quê?"

> **Resposta da IA (resumida):** "Como você está comparando a média de uma variável numérica (`Att`) entre duas categorias (lendário/normal), um **gráfico de barras** é o mais indicado — ele facilita a comparação direta das médias. Um `boxplot` também seria uma opção, caso você queira visualizar a distribuição completa e não apenas a média."

A sugestão faz sentido e é exatamente o que aprendemos na Aula 06! Vamos construir o gráfico:

````python
# criando uma coluna categórica a partir do campo Legendary (1.0 = lendário, 0.0 = normal)
pokemons['Tipo Lendário'] = pokemons['Legendary'].apply(lambda x: 'Lendário' if x == 1.0 else 'Normal')

# comparando a média de ataque entre os dois grupos
sns.barplot(pokemons, x='Tipo Lendário', y='Att')

# adicionando título e rótulos dos eixos
plt.title('Média de Ataque: Lendários vs. Normais')
plt.xlabel('Tipo')
plt.ylabel('Ataque Médio')

# exibe o gráfico
plt.show()
````

Como esperado, os Pokémons lendários têm, em média, um ataque bem mais alto que os normais. A IA acertou a recomendação — mas quem validou se ela fazia sentido, construiu o código e interpretou o resultado foi você.

## Exemplo 2: Gerando Código a Partir de um Prompt

Agora vamos usar a IA para gerar um trecho de código Pandas inteiro, e não apenas uma sugestão de gráfico.

> **Prompt:** "No DataFrame `pokemons`, tenho as colunas `Generation` (geração, de 1.0 a 8.0) e `Legendary` (1.0 ou 0.0). Gere um código Pandas que calcule o percentual de Pokémons lendários em cada geração."

> **Resposta da IA:**
> ```python
> legendarios_por_geracao = pokemons.groupby('Generation')['Legendary'].mean() * 100
> legendarios_por_geracao.round(2)
> ```
> "Como `Legendary` já é 1.0 ou 0.0, a média de cada grupo equivale diretamente ao percentual de lendários."

O raciocínio da IA está correto: a média de uma coluna binária (0 ou 1) é matematicamente igual à proporção de valores 1. Vamos rodar o código no nosso notebook:

````python
# calculando o percentual de pokemons lendários por geração
legendarios_por_geracao = pokemons.groupby('Generation')['Legendary'].mean() * 100
legendarios_por_geracao.round(2)
````

O resultado mostra que a proporção de lendários não é uniforme entre as gerações — algumas concentram bem mais Pokémons lendários que outras.

## Exemplo 3: Usando IA para Corrigir um Erro

Erros de código também são uma ótima oportunidade para usar a IA. Suponha que você tentou calcular a média de todos os atributos numéricos por geração com o seguinte código:

```python
media_por_geracao = pokemons.groupby('Generation').mean()
```

E recebeu um erro parecido com este:

```
TypeError: agg function failed [how to aggregate should be inferred from an ambiguous grouping, since not all columns are numeric]
```

Em vez de tentar adivinhar sozinho, você pode colar o erro direto na IA:

> **Prompt:** "Recebi o erro `TypeError: ... not all columns are numeric` ao rodar `pokemons.groupby('Generation').mean()`. O que está errado e como eu corrijo?"

> **Resposta da IA:** "O erro acontece porque o seu DataFrame tem colunas de texto (como `Name` e `Type 1`), e o Pandas não consegue calcular a média delas. A partir da versão 2.0 do Pandas, você precisa indicar explicitamente que quer considerar apenas colunas numéricas, usando o parâmetro `numeric_only=True`."

Vamos aplicar a correção e conferir:

````python
# corrigindo o erro: numeric_only=True calcula a média apenas das colunas numéricas
media_por_geracao = pokemons.groupby('Generation').mean(numeric_only=True)
media_por_geracao[['Att', 'Def']].round(2)
````

> ⚠️ **Atenção:** a explicação da IA fez sentido e o código funcionou, mas repare que foi você quem confirmou isso rodando a célula. Nunca copie uma correção de IA para um notebook sem executá-la e conferir o resultado.

## Exemplo 4: Pedindo Interpretação de um Resultado

Por fim, a IA também pode ajudar a transformar um gráfico ou uma tabela em texto — útil quando você precisa comunicar um resultado para alguém que não vai abrir o notebook.

> **Prompt:** "Tenho um gráfico de barras mostrando que a média de ataque dos Pokémons lendários é bem maior que a dos normais. Escreva 2 frases resumindo esse achado para um relatório."

> **Resposta da IA:** "Pokémons lendários apresentam, em média, um ataque significativamente superior ao dos Pokémons comuns, reforçando seu papel como as unidades mais poderosas do jogo. Essa diferença deve ser considerada em qualquer análise de balanceamento que envolva essa categoria."

> 💡 **Dica:** peça sempre para a IA basear a interpretação **apenas** nos números e no gráfico que você forneceu. Sem esse cuidado, ela pode complementar a resposta com informações genéricas sobre o universo Pokémon que não vêm dos seus dados.

## Limitações e Cuidados

Antes de adotar a IA generativa no seu fluxo de trabalho, é importante conhecer suas limitações:

- **Alucinação** — a IA pode inventar nomes de colunas, funções ou até resultados que parecem plausíveis, mas não existem
- **Desatualização** — o modelo pode sugerir uma sintaxe antiga de uma biblioteca (o `numeric_only=True` do Exemplo 3 é um bom exemplo de mudança recente)
- **Privacidade** — nunca envie dados sensíveis ou confidenciais para ferramentas de IA públicas sem antes anonimizar ou obter autorização
- **Viés** — os modelos aprendem a partir de dados existentes e podem reproduzir vieses presentes neles
- **Falta de contexto de negócio** — a IA não conhece as particularidades do seu problema; apenas o que você descrever no prompt

> ⚠️ **Atenção:** o principal risco de usar IA generativa em análise de dados não é ela estar errada — é você não perceber que ela está errada. Sempre valide o código executando-o e confira se os resultados fazem sentido com o que você conhece do dataset.

## Boas Práticas

Para incorporar a IA generativa de forma responsável na sua rotina como Cientista de Dados:

- Sempre execute e confira o código gerado antes de usá-lo
- Descreva o contexto do dataset (colunas, tipos, objetivo) no prompt
- Refine o prompt em várias interações, em vez de esperar a resposta perfeita na primeira tentativa
- Use a IA para acelerar tarefas repetitivas, não para pular etapas de entendimento
- Ao compartilhar um notebook, deixe claro quais trechos tiveram apoio de IA
