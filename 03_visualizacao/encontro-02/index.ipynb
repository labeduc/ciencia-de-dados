> 💡 **Dica:** `limite = max(pokemons['HP'].max(), pokemons['Spe'].max()) * 1.05` — o fator `1.05` dá uma folga de 5% para nenhum ponto encostar na borda. A linha divisória fica em `limite / 2`.

```python
# gera o gráfico de dispersão base
sns.scatterplot(pokemons, x='HP', y='Spe')

# títulos e rótulos
plt.title('Quadrantes: HP X Velocidade')
plt.xlabel('Pontos de Vida (HP)')
plt.ylabel('Velocidade')

# limite calculado dinamicamente a partir dos dados, com 5% de folga
limite = max(pokemons['HP'].max(), pokemons['Spe'].max()) * 1.05
plt.xlim(0, limite)
plt.ylim(0, limite)

# linhas divisórias na metade do limite
meio = limite / 2
plt.axhline(y=meio, color='blue', linestyle='--', linewidth=1)
plt.axvline(x=meio, color='blue', linestyle='--', linewidth=1)

# rótulo de cada quadrante
plt.text(x=limite*0.10, y=limite*0.04, s='Lentos e Frágeis', color='blue')
plt.text(x=limite*0.70, y=limite*0.04, s='Resistentes', color='blue')
plt.text(x=limite*0.10, y=limite*0.93, s='Rápidos', color='blue')
plt.text(x=limite*0.75, y=limite*0.93, s='Élite', color='blue')

# mostra o gráfico
plt.show()
```

**Explicações adicionais:**

- Na Aula 08 usamos valores fixos (`0, 225`) para simplificar. Aqui demos o passo seguinte: **calcular tudo a partir dos dados**. Se o dataset mudar, o gráfico se ajusta sozinho — inclusive os textos, posicionados como percentuais do limite.
- O quadrante superior direito (`Élite`) é o equivalente ao quadrante dos "líderes" no Quadrante Mágico do Gartner que vimos na aula: alto em **ambas** as dimensões.

### Exercício 3 - Pizza com Moderação

Crie um **gráfico de pizza** mostrando a **proporção de pokemons por geração**, com percentuais nas fatias e fatias em ordem (Gen 1, Gen 2, ...).

**Funções úteis:** `groupby()`, `count()`, `sort_values()`, `plt.pie()` com `autopct=` e `counterclock=False`

```python
# Escreva seu código aqui — vamos resolver juntos!
```

> 💡 **Dica:** agrupe por `Generation Desc` contando a coluna `Number`, ordene pelo rótulo e use `plt.pie(valores, labels=rotulos, autopct="%.2f%%", counterclock=False)`.

#### Solução comentada

```python
# agrupamos por geração e contamos os pokemons
pokemons_gen = pokemons.groupby('Generation Desc')['Number'].count().reset_index()

# ordenamos para as fatias saírem na sequência Gen 1, Gen 2, ...
pokemons_gen.sort_values('Generation Desc', inplace=True)

# pizza com percentual em cada fatia, no sentido horário
plt.pie(pokemons_gen['Number'], labels=pokemons_gen['Generation Desc'],
        autopct="%.2f%%", counterclock=False)

plt.title('Distribuição de Pokemons por Geração')
plt.show()
```

**Explicações adicionais:**

- Lembre-se do critério da Aula 09: pizza funciona bem com **poucas categorias** (aqui, 8) e quando a **proporção do todo** é a pergunta. Para comparar valores entre categorias, barras são quase sempre melhores.
- A ordenação por `Generation Desc` é **textual** — funciona até a `Gen 9`, mas uma hipotética `Gen 10` apareceria entre `Gen 1` e `Gen 2`. Em casos assim, ordene pela coluna numérica original.

### Exercício 4 - Mapa de Rede: Combinações de Tipos

Crie um **mapa de rede** mostrando quais **tipos primários** (`Type 1`) se combinam com quais **tipos secundários** (`Type 2`):

1. Remova as linhas sem `Type 2` (lembra do Exercício 1 do Encontro 1?)
2. Conte as combinações e filtre as que ocorrem **mais de 10 vezes**
3. Monte o grafo com arestas de **largura proporcional** à contagem

**Bibliotecas e funções úteis:** `dropna()`, `networkx as nx`, `nx.Graph()`, `add_edge()`, `nx.spring_layout()`, `nx.draw()`

```python
# Escreva seu código aqui — vamos resolver juntos!
```

> 💡 **Dica:** depois do `groupby(['Type 1', 'Type 2'])['Number'].count().reset_index(name='Total')`, filtre com `query('Total > 10')`. Para popular o grafo sem `iterrows()`, percorra as colunas com `zip(combos['Type 1'], combos['Type 2'], combos['Total'])`.

#### Solução comentada

```python
import networkx as nx

# 1. remove pokemons sem tipo secundário e conta as combinações
combos = (pokemons.dropna(subset=['Type 2'])
                  .groupby(['Type 1', 'Type 2'])['Number']
                  .count()
                  .reset_index(name='Total'))

# 2. mantém apenas as combinações frequentes, para o grafo ficar legível
combos = combos.query('Total > 10')

# 3. monta o grafo: cada aresta liga um Type 1 a um Type 2, com peso
net = nx.Graph()
for t1, t2, total in zip(combos['Type 1'], combos['Type 2'], combos['Total']):
    net.add_edge(t1, t2, weight=total)

# calcula a posição dos nodos e a largura de cada aresta (peso / 10)
pos = nx.spring_layout(net, seed=3068, k=0.8, iterations=50)
larguras = [net[u][v]['weight'] / 10 for u, v in net.edges()]

# plota o grafo
plt.rcParams["figure.figsize"] = [12.00, 12.00]
nx.draw(net, pos=pos, with_labels=True, node_size=1500, node_color='lightblue',
        edgecolors='black', font_size=10, width=larguras)
plt.show()

# restaura o tamanho de figura padrão do encontro
plt.rcParams["figure.figsize"] = (9, 5)
```

**Explicações adicionais:**

- O `dropna(subset=['Type 2'])` é essencial: valores `NaN` virariam nodos inválidos no grafo. E o filtro `Total > 10` evita o efeito "novelo de lã" que vimos na Aula 11 — um grafo com arestas demais não comunica nada.
- Repare que os **nodos** (tipos) são criados automaticamente pelo `add_edge()` — só precisamos declarar as **arestas** com seus pesos.
- A largura proporcional (`weight / 10`) transforma o peso em informação visual: quanto mais grossa a linha, mais comum a combinação de tipos.

## Live Coding — Um Visual Plotly Completo, Camada a Camada

Agora o instrutor vai construir, junto com vocês, um gráfico interativo completo — exatamente como fizemos na Aula 16, camada a camada. Começamos pela versão "crua":

```python
import plotly.express as px

# preparando os dados: média de ataque e defesa por geração, no formato longo
pokemons_gb = pokemons.groupby('Generation Desc')[['Att', 'Def']].mean().reset_index()
tidy = pokemons_gb.melt(id_vars='Generation Desc', var_name='Atributo', value_name='Pontos')

# versão 1: o gráfico sai funcional, mas sem nenhum cuidado visual
fig = px.bar(tidy, x='Generation Desc', y='Pontos', color='Atributo', barmode='group')
fig.show(renderer='notebook')
```

Funciona, mas está longe de pronto para uma apresentação: título ausente, eixos com nomes técnicos, fundo poluído e tooltip em "linguagem de programador". Agora, as camadas de configuração — acompanhe a explicação de cada bloco:

```python
# versão 2: o mesmo gráfico, agora com as camadas de configuração
fig = px.bar(tidy, x='Generation Desc', y='Pontos', color='Atributo',
             barmode='group',
             text='Pontos', text_auto='.1f',   # valores sobre as barras
             # labels renomeia as colunas nos eixos, na legenda e na tooltip
             labels={'Generation Desc': 'Geração',
                     'Pontos': 'Média dos pontos',
                     'Atributo': 'Atributo'})

# dicionário de configuração do layout (Aula 16)
configuracao = {
    "title": {"text": "Média de Ataque e Defesa por Geração"},
    "legend": {"title": {"text": "Atributo"}},
    "yaxis": {"visible": False},          # os valores já estão nas barras
    "xaxis": {"title": "Geração"},
    "plot_bgcolor": "#FFFFFF",            # fundo limpo
}
fig.update_layout(configuracao)

fig.show(renderer='notebook')
```

**Explicações adicionais:**

- O parâmetro `labels={}` resolve três coisas de uma vez: nomes dos eixos, título da legenda e os textos da **tooltip** — passe o mouse sobre as barras e compare com a versão 1!
- Removemos o eixo Y (`"yaxis": {"visible": False}`) porque os valores já estão escritos sobre as barras — é a regra de ouro da Aula 10: **remova o ruído, mantenha a informação**.
- O `update_layout()` recebe um dicionário — a mesma estrutura aninhada que vimos no objeto `Figure` da Aula 16 (`data`, `layout`, `config`).
