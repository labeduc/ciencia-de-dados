## **Como o ChatGPT usa Banco de Dados para responder suas perguntas?**

Você já se perguntou o que acontece **por trás** quando você faz uma pergunta ao ChatGPT?

Muita gente acha que a IA "sabe tudo de cabeça" — mas não é bem assim. Em muitos sistemas modernos de IA, a resposta depende de **consultar um banco de dados** para buscar informações relevantes antes de gerar uma resposta. Esse processo se chama **RAG** (*Retrieval-Augmented Generation* = Geração Aumentada por Recuperação).

Veja o fluxo simplificado:

<img src="img/02_banner_como_funciona.png">

### Traduzindo para o dia a dia:

| **Etapa** | **O que acontece** | **Analogia** |
|---|---|---|
| **1. Parse** | A IA interpreta o que você perguntou | Como quando você lê uma pergunta na prova e entende o que estão pedindo |
| **2. Embedding** | A pergunta vira números (vetores) | Como traduzir uma frase para um "código secreto" que o computador entende |
| **3. Busca no Banco** | O sistema procura informações relevantes | Como pesquisar no Google, mas dentro de um banco de dados privado |

**Sintaxe**: 
```

FROM <nome_da_tabela>

```

**Exemplo**:

## **Cláusula WHERE**


A cláusula Where permite ao comando SQL passar condições para filtrar dados.

Sintaxe:
```
SELECT <lista_de_campos>
FROM <nome_da_tabela>
WHERE <filtros>
```

Exemplo:
```
SELECT codigo, nome
FROM clientes
WHERE estado = 'RS'

```

Exemplo de outros operadores utilizados com a cláusula WHERE:


| **Operador**            | **Exemplo**
| ----------------------- | -------------------- 
| =                       | Select * from clientes where estado = ‘RS’
| <>                      | Select * from clientes where estado <> ‘RS’
| OR                      | Select * from clientes where estado = ‘RS’ OR estado = ‘SP’
| AND                     | Select * from clientes where estado = ‘RS’ AND cidade = ‘Porto Alegre’
| ()                      | Select * from clientes where (estado = ‘RS’ AND cidade = ‘Canoas’) OR (cidade = ‘Gravataí’)
| LIKE                    | Select * from clientes where nome LIKE ‘%Santos’
| NOT LIKE                | Select * from clientes where nome NOT LIKE ‘%Santos’
| IN                      | Select * from clientes where estado IN (‘RS’,’SP’)
| NOT IN                  | Select * from clientes where estado NOT IN (‘RS’,’SP’)
| BETWEEN                 | Select * from clientes where codigo BETWEEN 1 AND 3
| NOT BETWEEN             | Select * from clientes where codigo NOT BETWEEN 1 AND 3

## **Cláusula GROUP BY**

A cláusula group by permite agrupamento dos dados em um comando SQL.

**Sintaxe:**



```
SELECT <lista_de_campos>
FROM <nome_da_tabela>
WHERE <filtros>
GROUP BY <lista_de_campos>
```

**Exemplo:**



```
SELECT count(*),estado
FROM clientes
GROUP BY estado
```

Alguns operadores utilizados em agrupamentos:


*   Count - Contar número de linhas retornadas em uma consulta
*   Max - Traz o maior valor dentre as linhas retornas em uma consulta
*   Min - Traz o menor valor dentre as linhas retornas em uma consulta



````python
# Mostrando o conteúdo da tabela clientes
#cursor = conn.execute("SELECT * FROM Clientes")
#for row in cursor:
#  print(row)
%%sql
SELECT * FROM Clientes
````

````python
# Mostrando a quantidade de clientes
#cursor = conn.execute("SELECT count(*) FROM clientes")
#for row in cursor:
#  print(row)
%%sql
SELECT count(*) FROM clientes
````

````python
# Mostrando a quantidade de clientes por estado
#cursor = conn.execute("SELECT count(*),estado FROM clientes GROUP BY estado")
#for row in cursor:
#  print(row)
%%sql
SELECT count(*),estado FROM clientes GROUP BY estado
````

## **Cláusula HAVING**

A cláusula HAVING determina uma condição de busca em conjunto com o GROUP BY.

**Sintaxe:**



```
SELECT <lista_de_campos>
FROM <nome_da_tabela>
WHERE <filtros>
GROUP BY <lista_de_campos>
HAVING <condicao>
```

**Exemplo:**



```
SELECT count(*),estado
FROM clientes
GROUP BY estado
HAVING count(*) > 1
```



````python
# Mostrando o conteúdo da tabela clientes
#cursor = conn.execute("SELECT * FROM Clientes")
#for row in cursor:
#  print(row)
%%sql
SELECT * FROM Clientes
````

````python
# Mostrando a quantidade de clientes por estado onde tenhamos mais que um cliente por estado
#cursor = conn.execute("SELECT count(*),estado FROM clientes GROUP BY estado HAVING count(*) > 1")
#for row in cursor:
#  print(row)
%%sql
SELECT count(*),estado FROM clientes GROUP BY estado HAVING count(*) > 1
````
