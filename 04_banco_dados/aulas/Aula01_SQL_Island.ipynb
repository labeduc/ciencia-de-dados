```python
# Conectar no banco de dados - sql_island.db
conn = sqlite3.connect('sql_island.db')
```

```python
# Mostrando o conteúdo da tabela aldeia
cursor_aldeia = conn.execute("select * from aldeia")
for row in cursor_aldeia:
  print(row)
```
