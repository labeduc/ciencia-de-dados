## **O que é Modelagem de Dados?** 🤔

Modelagem de dados é o processo de **planejar e desenhar a estrutura** de um banco de dados antes de criá-lo.

Pense assim:

| **Analogia** | **Sem modelagem** | **Com modelagem** |
|---|---|---|
| 🏠 Construir uma casa | Começa colocando tijolos sem planta | Primeiro faz a planta, depois constrói |
| 🎮 Criar um jogo | Começa programando sem game design | Primeiro planeja fases, personagens e regras |
| 📱 Desenvolver um app | Sai codando direto | Primeiro desenha as telas e o fluxo do usuário |
| 🗄️ Banco de dados | Cria tabelas aleatórias | Primeiro modela entidades, atributos e relacionamentos |

**Modelar é pensar antes de fazer.** E no mundo dos dados, isso evita retrabalho, erros e bagunça.

## **Por que modelagem importa (especialmente na era da IA)?** 🤖

Lembra que na Aula 01 vimos que **dados são o combustível da IA**? Pois é — se o combustível estiver **sujo ou desorganizado**, a IA não funciona direito.

### Exemplos reais:

| **Situação** | **Banco mal modelado** | **Banco bem modelado** |
|---|---|---|
| 🎵 Spotify recomendando música | Não sabe diferenciar artista de banda | Tabela `Artista` separada de `Banda`, com relacionamento claro |
| 🛒 Shopee mostrando produtos | Produtos duplicados, preços errados | Cada produto único, com categoria, preço e estoque organizados |
| 💬 ChatGPT buscando informação | Documentos repetidos, sem categorização | Documentos indexados, categorizados e sem duplicação |
| 🎮 Steam salvando progresso | Conquistas misturadas entre jogos | Tabela `Jogo`, tabela `Conquista`, tabela `Jogador` — tudo separado e conectado |

> **Regra de ouro:** quanto melhor a modelagem, melhor a qualidade dos dados → melhor o resultado da IA. 🏆

---

### A IA precisa de dados organizados

Quando você treina um modelo de IA ou faz RAG (como o ChatGPT busca informações em bancos de dados), os dados precisam estar:

1. **Sem duplicação** — senão a IA dá respostas repetidas ou enviesadas
2. **Com relacionamentos claros** — senão a IA não consegue conectar informações
3. **Com tipos corretos** — senão a IA confunde texto com número, data com nome
4. **Com integridade** — senão a IA trabalha com dados "fantasma" (referências quebradas)

Tudo isso é garantido por uma **boa modelagem**.

## **Os 3 conceitos fundamentais da modelagem** 🧩

Toda modelagem de dados gira em torno de **3 conceitos**:

### 1. Entidade (= Tabela)

Uma **entidade** é qualquer "coisa" do mundo real que queremos armazenar informações sobre.

| **Mundo real** | **Entidade no banco** |
|---|---|
| Um aluno da escola | `Aluno` |
| Um jogo no Steam | `Jogo` |
| Uma música no Spotify | `Musica` |
| Um post no Instagram | `Post` |
| Um pedido na Shopee | `Pedido` |

Cada entidade vira uma **tabela** no banco de dados.

---

### 2. Atributo (= Coluna)

Um **atributo** é uma característica da entidade — cada detalhe que queremos guardar.

| **Entidade** | **Atributos** |
|---|---|
| `Aluno` | nome, idade, turma, email |
| `Jogo` | titulo, genero, preco, plataforma |
| `Musica` | titulo, artista, duracao, genero |
| `Post` | texto, data, curtidas, usuario |

Cada atributo vira uma **coluna** na tabela.

---

### 3. Relacionamento (= Chave Estrangeira)

Um **relacionamento** mostra como as entidades se conectam entre si.

| **Relacionamento** | **Exemplo** |
|---|---|
| Um **Aluno** pertence a uma **Turma** | `Aluno.turma_id` → `Turma.id` |
| Um **Jogador** compra vários **Jogos** | `Compra.jogador_id` → `Jogador.id` |
| Uma **Música** pertence a um **Artista** | `Musica.artista_id` → `Artista.id` |
| Um **Post** pertence a um **Usuário** | `Post.usuario_id` → `Usuario.id` |

Cada relacionamento é implementado com **chave estrangeira** (`FOREIGN KEY`) — que você já viu na Aula 02!

## **Tipos de Relacionamento** 🔗

Existem 3 tipos de relacionamento entre entidades:

### 1:1 (Um para Um)
Cada registro de uma tabela se relaciona com **exatamente um** registro de outra tabela.

> 🎮 Um **Jogador** tem **um** Perfil. Um **Perfil** pertence a **um** Jogador.

### 1:N (Um para Muitos)
Um registro de uma tabela se relaciona com **vários** registros de outra tabela. **É o mais comum!**

> 🎵 Um **Artista** tem **várias** Músicas. Cada **Música** pertence a **um** Artista.
>
> 📱 Um **Usuário** faz **vários** Posts. Cada **Post** pertence a **um** Usuário.

### N:N (Muitos para Muitos)
Vários registros de uma tabela se relacionam com **vários** registros de outra tabela.

> 🎓 Um **Aluno** cursa **várias** Disciplinas. Uma **Disciplina** tem **vários** Alunos.
>
> 🎮 Um **Jogador** joga **vários** Jogos. Um **Jogo** é jogado por **vários** Jogadores.

⚠️ **Atenção:** relacionamentos N:N precisam de uma **tabela intermediária** (tabela associativa) para funcionar!

Exemplo: `Aluno_Disciplina` com colunas `aluno_id` e `disciplina_id`.

## **Os 3 Níveis de Modelagem** 📊

A modelagem de dados acontece em **3 etapas**, do mais abstrato ao mais concreto:

| **Nível** | **O que é** | **Analogia** | **Quem faz** |
|---|---|---|---|
| **Conceitual** | Visão geral das entidades e relacionamentos | A planta da casa (sem medidas) | Analista / Cliente |
| **Lógico** | Define atributos, tipos e chaves | A planta com medidas e materiais | Analista / Projetista |
| **Físico** | O `CREATE TABLE` real no banco | A casa sendo construída | DBA / Desenvolvedor |

---

### Na prática:

**1️⃣ Modelo Conceitual** — "O quê?"
- Identifica as **entidades** (Aluno, Turma, Disciplina)
- Identifica os **relacionamentos** (Aluno *cursa* Disciplina)
- Não se preocupa com tipos de dados nem com SQL

**2️⃣ Modelo Lógico** — "Como?"
- Define os **atributos** de cada entidade (nome, idade, email)
- Define os **tipos de dados** (VARCHAR, INT, DATE)
- Define **chaves primárias** e **chaves estrangeiras**
- Independente do SGBD (funciona para SQLite, MySQL, PostgreSQL...)

**3️⃣ Modelo Físico** — "Onde e exatamente como?"
- É o código SQL real (`CREATE TABLE`)
- Específico para o SGBD escolhido
- Inclui índices, constraints e otimizações
- É o que você já fez na Aula 02! 🎉

## **Diagrama Entidade-Relacionamento (ER)** 📐

O **Diagrama ER** é a ferramenta principal da modelagem. É como o "mapa" do banco de dados.

Os elementos do diagrama são:

| **Elemento** | **Símbolo** | **Significado** |
|---|---|---|
| Entidade | 🟦 Retângulo | Uma tabela (ex: `Aluno`, `Jogo`) |
| Atributo | ⚪ Elipse/Oval | Uma coluna (ex: `nome`, `idade`) |
| Relacionamento | 🔷 Losango | Conexão entre entidades (ex: "cursa", "compra") |
| Chave primária | ⚪ Sublinhado | Atributo que identifica cada registro (ex: `id`) |

Vamos modelar e criar um banco de dados de uma **rede social de games** — tipo uma mistura de Steam + Discord.

O sistema precisa armazenar:

- **Jogos** (os games disponíveis)

- **Partidas** (histórico de partidas jogadas)

Primeiro, vamos preparar o ambiente:

```python
# Importar a biblioteca sqlite3
import sqlite3

# Conectar (cria o banco se não existir)
conn = sqlite3.connect('game_social.db')

# Habilitar suporte a FOREIGN KEY no SQLite
conn.execute('PRAGMA foreign_keys = ON')

# Carregar extensão SQL para usar %sql e %%sql
%load_ext sql
%sql sqlite:///game_social.db
```

### 🔵 Passo 1 — Modelo Conceitual

Identificando as **entidades** e **relacionamentos**:

<img src="img/03_modelagem_conceitual.png">

**Leitura:**
- Um **Jogador** joga **vários** Jogos. Um **Jogo** é jogado por **vários** Jogadores → **N:N** (precisa de tabela intermediária!)
- Um **Jogo** tem **várias** Conquistas. Uma **Conquista** pertence a **um** Jogo → **1:N**

### 🟢 Passo 2 — Modelo Lógico

Agora definimos os **atributos**, **tipos de dados** e **chaves**:

| **Entidade** | **Atributo** | **Tipo** | **Chave** |
|---|---|---|---|
| `Jogador` | id | INT | PK |
| | nickname | VARCHAR(50) | UNIQUE, NOT NULL |
| | email | VARCHAR(100) | UNIQUE, NOT NULL |
| | nivel | INT | DEFAULT 1 |
| | data_cadastro | DATE | NOT NULL |
| `Jogo` | id | INT | PK |
| | titulo | VARCHAR(100) | NOT NULL |
| | genero | VARCHAR(30) | NOT NULL |
| | classificacao | VARCHAR(5) | CHECK (L, 10, 12, 14, 16, 18) |
| | preco | REAL | CHECK (>= 0) |
| `Conquista` | id | INT | PK |
| | nome | VARCHAR(100) | NOT NULL |
| | descricao | VARCHAR(255) | |
| | pontos | INT | DEFAULT 10 |
| | jogo_id | INT | FK → Jogo.id |
| `Partida` | id | INT | PK |
| | jogador_id | INT | FK → Jogador.id |
| | jogo_id | INT | FK → Jogo.id |
| | data_partida | DATETIME | NOT NULL |
| | duracao_min | INT | CHECK (> 0) |
| | pontuacao | INT | DEFAULT 0 |

A tabela `Partida` é a **tabela intermediária** do relacionamento N:N entre Jogador e Jogo!

### 🔴 Passo 3 — Modelo Físico (CREATE TABLE)

Agora transformamos o modelo lógico em SQL real. Essa parte vocês já sabem da Aula 02! 💪

```python
%%sql
CREATE TABLE Jogador (
    id INTEGER PRIMARY KEY,
    nickname VARCHAR(50) UNIQUE NOT NULL,
    email VARCHAR(100) UNIQUE NOT NULL,
    nivel INTEGER DEFAULT 1,
    data_cadastro DATE NOT NULL
);
```

```python
%%sql
CREATE TABLE Jogo (
    id INTEGER PRIMARY KEY,
    titulo VARCHAR(100) NOT NULL,
    genero VARCHAR(30) NOT NULL,
    classificacao VARCHAR(5) CHECK(classificacao IN ('L', '10', '12', '14', '16', '18')),
    preco REAL CHECK(preco >= 0)
);
```

```python
%%sql
CREATE TABLE Conquista (
    id INTEGER PRIMARY KEY,
    nome VARCHAR(100) NOT NULL,
    descricao VARCHAR(255),
    pontos INTEGER DEFAULT 10,
    jogo_id INTEGER NOT NULL,
    FOREIGN KEY (jogo_id) REFERENCES Jogo(id)
);
```

```python
%%sql
CREATE TABLE Partida (
    id INTEGER PRIMARY KEY,
    jogador_id INTEGER NOT NULL,
    jogo_id INTEGER NOT NULL,
    data_partida DATETIME NOT NULL,
    duracao_min INTEGER CHECK(duracao_min > 0),
    pontuacao INTEGER DEFAULT 0,
    FOREIGN KEY (jogador_id) REFERENCES Jogador(id),
    FOREIGN KEY (jogo_id) REFERENCES Jogo(id)
);
```

## **Consultando o banco modelado** 🔍

Agora vem a parte legal: com um banco bem modelado, as consultas ficam **poderosas e fáceis de entender**!

### Quais jogos cada jogador jogou?

```python
%%sql
SELECT j.nickname, g.titulo, p.pontuacao, p.duracao_min
FROM Partida p
JOIN Jogador j ON p.jogador_id = j.id
JOIN Jogo g ON p.jogo_id = g.id
ORDER BY j.nickname, p.data_partida;
```

### Quantas partidas cada jogador jogou de cada jogo?

```python
%%sql
SELECT j.nickname, g.titulo, COUNT(*) as total_partidas, SUM(p.pontuacao) as pontos_total
FROM Partida p
JOIN Jogador j ON p.jogador_id = j.id
JOIN Jogo g ON p.jogo_id = g.id
GROUP BY j.nickname, g.titulo
ORDER BY pontos_total DESC;
```

### Quais conquistas existem para cada jogo?

```python
%%sql
SELECT g.titulo, c.nome as conquista, c.descricao, c.pontos
FROM Conquista c
JOIN Jogo g ON c.jogo_id = g.id
ORDER BY g.titulo, c.pontos DESC;
```

## **Normalização: a arte de evitar bagunça** 🧹

Normalização é o processo de **organizar as tabelas** para evitar problemas como:

- **Dados repetidos** (redundância)
- **Informações inconsistentes** (anomalias)
- **Desperdício de espaço**

### Exemplo: banco mal modelado (sem normalização)

Imagine guardar partidas assim, tudo em **uma tabela só**:

| partida_id | jogador_nick | jogador_email | jogo_titulo | jogo_genero | jogo_preco | pontuacao |
|---|---|---|---|---|---|---|
| 1 | xXDarkSlayerXx | dark@email.com | Valorant | FPS | 0.0 | 2500 |
| 2 | xXDarkSlayerXx | dark@email.com | Fortnite | Battle Royale | 0.0 | 1800 |
| 3 | NinjaDoPS5 | ninja@email.com | Fortnite | Battle Royale | 0.0 | 950 |

**Problemas:**
- O nome, email e gênero do jogo se **repetem** em cada linha 😩
- Se o jogador mudar de email, precisa **atualizar em todas as linhas** 😰
- Se deletar todas as partidas de um jogo, **perde as informações do jogo** 😱

### Banco bem modelado (normalizado)

É exatamente o que fizemos acima: cada informação fica em **uma tabela só** e as conexões são feitas por **chaves**. Sem repetição, sem inconsistência, sem perda de dados! ✅

## **Boas Práticas de Modelagem** ✨

Para fechar, aqui vão as regras de ouro que todo bom modelador segue:

| **#** | **Regra** | **Exemplo** |
|---|---|---|
| 1 | Cada tabela deve ter uma **chave primária** | `id INTEGER PRIMARY KEY` |
| 2 | Nomes de tabelas no **singular** | `Jogador` (não `Jogadores`) |
| 3 | Nomes de colunas em **snake_case** | `data_cadastro` (não `dataCadastro`) |
| 4 | Evitar **dados repetidos** entre tabelas | Usar FK em vez de copiar dados |
| 5 | Usar **tipos de dados adequados** | INT para números, VARCHAR para texto, DATE para datas |
| 6 | Definir **constraints** para proteger os dados | NOT NULL, CHECK, UNIQUE, FK |
| 7 | Relacionamentos N:N usam **tabela intermediária** | `Partida` conecta `Jogador` e `Jogo` |
| 8 | **Documentar** o modelo (diagrama ER) | Desenhar antes de codar |
