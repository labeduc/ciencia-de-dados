Nas aulas anteriores, você aprendeu a trabalhar com **bancos de dados relacionais** — aqueles de tabelas, colunas, linhas, `SELECT`, `JOIN` e tudo mais.

Mas... e se eu te dissesse que **nem todo dado do mundo cabe numa tabela**?

>
> Colocar tudo isso em tabelas com colunas fixas começa a ficar... complicado. 😅

É aí que entra o **NoSQL** — uma forma diferente de pensar e organizar dados.

## **SQL vs NoSQL: qual a diferença?** ⚔️

| | **SQL (Relacional)** | **NoSQL (Não-Relacional)** |
|---|---|---|
| **Estrutura** | Tabelas com colunas fixas | Documentos, chave-valor, grafos... |
| **Esquema** | Rígido — precisa definir colunas antes | Flexível — cada documento pode ter campos diferentes |
| **Linguagem** | SQL (`SELECT`, `INSERT`...) | Varia por banco (MongoDB usa JSON) |
| **Relacionamentos** | `JOIN` entre tabelas | Dados aninhados dentro do documento |
| **Exemplos** | SQLite, MySQL, PostgreSQL, Oracle | MongoDB, Redis, Cassandra, Neo4j |
| **Quando usar** | Dados estruturados e relacionados | Dados flexíveis, grande volume, alta velocidade |

| 📱 Instagram / TikTok | | ✅ Posts, stories, feeds em tempo real |
| 🎮 Steam / Discord | ✅ Compras e contas | ✅ Chat, status online, atividades |

> 💡 **A maioria das empresas usa os DOIS!** SQL para dados que precisam de consistência e NoSQL para dados que precisam de flexibilidade e velocidade.

## **Tipos de Banco NoSQL** 🗂️

NoSQL não é um tipo só — existem **4 categorias principais**:

| **Tipo** | **Como funciona** | **Exemplo** | **Usado para** |
|---|---|---|---|
| 📄 **Documentos** | Armazena dados como documentos JSON/BSON | **MongoDB** | Catálogos, perfis, conteúdo flexível |
| 🔑 **Chave-Valor** | Cada dado tem uma chave única e um valor | **Redis** | Cache, sessões, filas rápidas |
| 📊 **Colunar** | Organiza por colunas em vez de linhas | **Cassandra** | Big Data, séries temporais |
| 🕸️ **Grafos** | Armazena nós e conexões entre eles | **Neo4j** | Redes sociais, recomendações |

Nesta aula vamos focar no tipo mais popular: **Documentos**, usando o **MongoDB**! 🍃

## **MongoDB: o banco de documentos mais popular do mundo** 🍃

O MongoDB armazena dados como **documentos JSON** — que são basicamente **dicionários Python**!

### Vocabulário: SQL → MongoDB

| **SQL** | **MongoDB** | **Exemplo** |
|---|---|---|
| Banco de dados | Banco de dados | `game_social` |
| Tabela | **Coleção** (collection) | `jogadores` |
| Linha | **Documento** (document) | `{"nickname": "xXDarkSlayerXx", "nivel": 42}` |
| Coluna | **Campo** (field) | `nickname`, `nivel` |
| `PRIMARY KEY` | `_id` (gerado automaticamente) | |
| `SELECT` | `find()` | |
| `INSERT` | `insert_one()` / `insert_many()` | |
| `UPDATE` | `update_one()` / `update_many()` | |
| `DELETE` | `delete_one()` / `delete_many()` | |

---

### O poder do documento: dados aninhados!

No SQL, para guardar um jogador com seus jogos favoritos, você precisaria de **2 tabelas + JOIN**.

No MongoDB, tudo fica **dentro de um documento só**:

```json
{
    "nickname": "xXDarkSlayerXx",
    "nivel": 42,
    "email": "dark@email.com",
    "jogos_favoritos": ["Valorant", "Fortnite", "Minecraft"],
    "conquistas": [
        {"nome": "Headshot!", "jogo": "Valorant", "data": "2024-12-01"},
        {"nome": "Primeira Vitória", "jogo": "Fortnite", "data": "2024-12-02"}
    ],
    "config": {
        "tema": "escuro",
        "idioma": "pt-BR",
        "notificacoes": true
    }
}
```

Percebeu? **Listas dentro de listas, objetos dentro de objetos** — sem precisar de tabela intermediária, sem `JOIN`, sem complicação! 🎉

## **Inserindo documentos** 📝

No MongoDB, inserir dados é como criar um **dicionário Python** e salvar na coleção.

### `insert_one()` — inserir um documento

````python
# Inserir um jogador — repare como é parecido com um dicionário Python!
jogador1 = {
    "nickname": "xXDarkSlayerXx",
    "email": "dark@email.com",
    "nivel": 42,
    "jogos_favoritos": ["Valorant", "Fortnite", "Minecraft"],
    "conquistas": [
        {"nome": "Headshot!", "jogo": "Valorant", "pontos": 25},
        {"nome": "Primeira Vitória", "jogo": "Fortnite", "pontos": 10}
    ],
    "config": {
        "tema": "escuro",
        "idioma": "pt-BR",
        "notificacoes": True
    }
}

resultado = jogadores.insert_one(jogador1)
print(f"Jogador inserido com _id: {resultado.inserted_id}")
````

### `insert_many()` — inserir vários documentos de uma vez

Repare que cada jogador pode ter campos **diferentes** — um tem `clan`, outro não. Um tem 5 jogos favoritos, outro tem 2. **No NoSQL isso é normal!**

````python
outros_jogadores = [
    {
        "nickname": "NinjaDoPS5",
        "email": "ninja@email.com",
        "nivel": 28,
        "jogos_favoritos": ["Fortnite", "FIFA 25"],
        "conquistas": [
            {"nome": "Hat-Trick", "jogo": "FIFA 25", "pontos": 30}
        ],
        "config": {"tema": "claro", "idioma": "pt-BR", "notificacoes": False}
    },
    {
        "nickname": "GamerGirl_BR",
        "email": "gamer@email.com",
        "nivel": 55,
        "jogos_favoritos": ["Minecraft", "God of War", "Valorant", "Stardew Valley", "The Last of Us"],
        "conquistas": [
            {"nome": "Construtor Mestre", "jogo": "Minecraft", "pontos": 50},
            {"nome": "The End", "jogo": "Minecraft", "pontos": 100},
            {"nome": "Ragnarök", "jogo": "God of War", "pontos": 100}
        ],
        "clan": "BRForce",
        "config": {"tema": "rosa", "idioma": "pt-BR", "notificacoes": True}
    },
    {
        "nickname": "BotDeSuporte",
        "email": "bot@email.com",
        "nivel": 15,
        "jogos_favoritos": ["FIFA 25"],
        "conquistas": [],
        "config": {"tema": "escuro", "idioma": "en-US", "notificacoes": True}
    },
    {
        "nickname": "ProPlayerZero",
        "email": "pro@email.com",
        "nivel": 99,
        "jogos_favoritos": ["Valorant", "Minecraft", "CS2"],
        "conquistas": [
            {"nome": "Ace!", "jogo": "Valorant", "pontos": 75},
            {"nome": "Headshot!", "jogo": "Valorant", "pontos": 25},
            {"nome": "The End", "jogo": "Minecraft", "pontos": 100}
        ],
        "clan": "eSports_BR",
        "bio": "Streamer e competidor profissional 🏆",
        "config": {"tema": "escuro", "idioma": "pt-BR", "notificacoes": False, "stream_mode": True}
    }
]

resultado = jogadores.insert_many(outros_jogadores)
print(f"Inseridos {len(resultado.inserted_ids)} jogadores!")
````

## **Consultando documentos** 🔍

No MongoDB, as consultas são feitas com `find()` e `find_one()` — usando **dicionários como filtro** em vez de SQL.

| **SQL** | **MongoDB** |
|---|---|
| `SELECT * FROM jogadores` | `jogadores.find()` |
| `SELECT * FROM jogadores WHERE nivel > 30` | `jogadores.find({"nivel": {"$gt": 30}})` |
| `SELECT * FROM jogadores WHERE nickname = 'NinjaDoPS5'` | `jogadores.find_one({"nickname": "NinjaDoPS5"})` |
| `SELECT * FROM jogadores ORDER BY nivel DESC LIMIT 1` | `jogadores.find().sort("nivel", -1).limit(1)` |

### Listar todos os jogadores

````python
# find() retorna todos os documentos da coleção
for jogador in jogadores.find():
    print(f"🎮 {jogador['nickname']} — Nível {jogador['nivel']} — Jogos: {', '.join(jogador['jogos_favoritos'])}")
````

### Buscar um jogador específico

````python
# find_one() retorna o primeiro documento que bate com o filtro
jogador = jogadores.find_one({"nickname": "GamerGirl_BR"})

print(f"Nickname: {jogador['nickname']}")
print(f"Nível: {jogador['nivel']}")
print(f"Clan: {jogador.get('clan', 'Sem clan')}")
print(f"Jogos favoritos: {jogador['jogos_favoritos']}")
print(f"Conquistas: {len(jogador['conquistas'])}")
for c in jogador['conquistas']:
    print(f"  🏆 {c['nome']} ({c['jogo']}) — {c['pontos']} pts")
````

### Operadores de filtro (equivalente ao WHERE do SQL)

| **SQL** | **MongoDB** | **Significado** |
|---|---|---|
| `=` | `{"campo": valor}` | Igual |
| `>` | `{"campo": {"$gt": valor}}` | Maior que |
| `>=` | `{"campo": {"$gte": valor}}` | Maior ou igual |
| `<` | `{"campo": {"$lt": valor}}` | Menor que |
| `<=` | `{"campo": {"$lte": valor}}` | Menor ou igual |
| `<>` ou `!=` | `{"campo": {"$ne": valor}}` | Diferente |
| `IN (...)` | `{"campo": {"$in": [...]}}` | Está na lista |
| `AND` | Múltiplos campos no dicionário | E lógico |
| `OR` | `{"$or": [{...}, {...}]}` | OU lógico |

### Jogadores com nível acima de 30

````python
# Equivalente a: SELECT * FROM jogadores WHERE nivel > 30
for jogador in jogadores.find({"nivel": {"$gt": 30}}):
    print(f"🎮 {jogador['nickname']} — Nível {jogador['nivel']}")
````

### Jogadores ordenados por nível (do maior para o menor)

````python
# Equivalente a: SELECT * FROM jogadores ORDER BY nivel DESC
# sort(campo, direção) → 1 = crescente, -1 = decrescente
for jogador in jogadores.find().sort("nivel", -1):
    print(f"🎮 {jogador['nickname']} — Nível {jogador['nivel']}")
````

### 🔥 Superpoder do NoSQL: buscar DENTRO de listas!

No SQL, para saber quais jogadores gostam de Valorant, você precisaria de um `JOIN` com tabela intermediária. No MongoDB, basta:

````python
# Quem tem "Valorant" na lista de jogos favoritos?
for jogador in jogadores.find({"jogos_favoritos": "Valorant"}):
    print(f"🎮 {jogador['nickname']} curte Valorant!")
````

### Buscar dentro de objetos aninhados

Também dá para buscar dentro de subdocumentos usando **notação de ponto** (dot notation):

````python
# Quem usa tema escuro?
for jogador in jogadores.find({"config.tema": "escuro"}):
    print(f"🌙 {jogador['nickname']} usa tema escuro")
````

## **Atualizando documentos** ✏️

No MongoDB, o update usa **operadores especiais**:

| **Operador** | **O que faz** | **Exemplo** |
|---|---|---|
| `$set` | Define/atualiza um campo | `{"$set": {"nivel": 50}}` |
| `$inc` | Incrementa um valor numérico | `{"$inc": {"nivel": 5}}` |
| `$push` | Adiciona item a uma lista | `{"$push": {"jogos_favoritos": "CS2"}}` |
| `$pull` | Remove item de uma lista | `{"$pull": {"jogos_favoritos": "FIFA 25"}}` |
| `$unset` | Remove um campo do documento | `{"$unset": {"bio": ""}}` |

````python
# Subir o nível do BotDeSuporte para 20
jogadores.update_one(
    {"nickname": "BotDeSuporte"},
    {"$set": {"nivel": 20}}
)

# Verificar
bot = jogadores.find_one({"nickname": "BotDeSuporte"})
print(f"BotDeSuporte agora é nível {bot['nivel']}!")
````

````python
# Adicionar um novo jogo favorito ao NinjaDoPS5
jogadores.update_one(
    {"nickname": "NinjaDoPS5"},
    {"$push": {"jogos_favoritos": "Valorant"}}
)

# Verificar
ninja = jogadores.find_one({"nickname": "NinjaDoPS5"})
print(f"Jogos do NinjaDoPS5: {ninja['jogos_favoritos']}")
````

````python
# Subir 10 níveis para TODOS os jogadores de uma vez!
jogadores.update_many(
    {},
    {"$inc": {"nivel": 10}}
)

# Verificar
for j in jogadores.find():
    print(f"🎮 {j['nickname']} — Nível {j['nivel']}")
````

## **Deletando documentos** 🗑️

| **Método** | **O que faz** |
|---|---|
| `delete_one({filtro})` | Deleta o primeiro documento que bate com o filtro |
| `delete_many({filtro})` | Deleta todos os documentos que batem com o filtro |
| `delete_many({})` | ⚠️ Deleta TODOS os documentos da coleção |

````python
# Deletar o BotDeSuporte (ele era só um bot mesmo 🤖)
resultado = jogadores.delete_one({"nickname": "BotDeSuporte"})
print(f"Documentos deletados: {resultado.deleted_count}")

# Verificar que ele foi embora
print("\nJogadores restantes:")
for j in jogadores.find():
    print(f"  🎮 {j['nickname']}")
````

## **Comparação lado a lado: SQL vs MongoDB** 🔄

Vamos ver as mesmas operações nas duas linguagens:

| **Operação** | **SQL (SQLite)** | **MongoDB (mongita)** |
|---|---|---|
| Criar tabela/coleção | `CREATE TABLE Jogador (...)` | Automático no primeiro `insert` |
| Inserir | `INSERT INTO Jogador VALUES (...)` | `jogadores.insert_one({...})` |
| Buscar todos | `SELECT * FROM Jogador` | `jogadores.find()` |
| Buscar com filtro | `SELECT * FROM Jogador WHERE nivel > 30` | `jogadores.find({"nivel": {"$gt": 30}})` |
| Ordenar | `SELECT * FROM Jogador ORDER BY nivel DESC` | `jogadores.find().sort("nivel", -1)` |
| Atualizar | `UPDATE Jogador SET nivel = 50 WHERE nickname = 'x'` | `jogadores.update_one({"nickname": "x"}, {"$set": {"nivel": 50}})` |
| Deletar | `DELETE FROM Jogador WHERE nickname = 'x'` | `jogadores.delete_one({"nickname": "x"})` |
| Contar | `SELECT COUNT(*) FROM Jogador` | `jogadores.count_documents({})` |
| Buscar em lista | `JOIN + tabela intermediária` | `jogadores.find({"jogos_favoritos": "Valorant"})` |

> 💡 Percebeu? No MongoDB, **não existe `CREATE TABLE`** — a coleção é criada automaticamente quando você insere o primeiro documento!

## **Quando usar SQL e quando usar NoSQL?** 🤔

| **Use SQL quando...** | **Use NoSQL quando...** |
|---|---|
| Os dados têm **estrutura fixa** (todos iguais) | Os dados são **flexíveis** (cada registro pode ser diferente) |
| Precisa de **transações** seguras (ex: dinheiro) | Precisa de **velocidade** e **escala** massiva |
| Tem muitos **relacionamentos** entre dados | Os dados são **auto-contidos** (tudo num documento só) |
| Precisa de **integridade** rígida (constraints, FK) | Precisa armazenar **listas**, **objetos aninhados**, **dados variáveis** |
| Ex: banco do Itaú, sistema de matrícula | Ex: feed do TikTok, chat do Discord, catálogo do Steam |

---

### E a IA?

A IA moderna usa **os dois**:

- **SQL** → para dados estruturados de treinamento (tabelas de clientes, produtos, vendas)
- **NoSQL (MongoDB)** → para armazenar documentos de contexto no **RAG** (Retrieval-Augmented Generation)
- **NoSQL (Redis)** → para cache de respostas e sessões de chat
- **NoSQL (Vetorial)** → para busca semântica (como o ChatGPT encontra informações relevantes)
