Ao final, você deverá conseguir:

## Nosso ciclo de trabalho

# **Entender → Esboçar → Explicar → Criticar → Decidir → Revisar**

1. **Entender:** quem usa e qual tarefa precisa realizar?
2. **Esboçar:** como as informações serão organizadas?
3. **Explicar:** o que cada área mostra e por que é necessária?
4. **Criticar:** que problemas de usabilidade podem existir?
5. **Decidir:** quais sugestões realmente ajudam?
6. **Revisar:** qual mudança mínima melhora o protótipo?

## 3. Três princípios de UX

### Hierarquia visual
A informação mais importante deve ser encontrada primeiro. KPIs ficam no topo e o gráfico principal recebe mais espaço.

### Agrupamento lógico
Elementos relacionados ficam próximos: filtros com filtros, KPIs com KPIs e gráficos complementares na mesma região.

### Simplicidade
Cada elemento deve justificar o espaço que ocupa. Evite gráficos repetidos, decoração excessiva e rótulos confusos.

### Acessibilidade básica
Não comunique significado somente por cor. Use rótulos, unidades, texto legível e espaço suficiente entre elementos.

## 4. Anatomia do wireframe

```text
┌────────────────────────────────────────────────────┐
│ Vendas e desperdício             Filtros           │
├──────────────┬──────────────┬──────────────────────┤
│ Receita      │ Vendidos     │ Desperdício          │
├──────────────┴──────────────┴──────────────────────┤
│                                                    │
│ Gráfico principal: evolução por dia                │
│                                                    │
├──────────────────────────┬─────────────────────────┤
│ Produtos mais vendidos   │ Fonte e atualização     │
└──────────────────────────┴─────────────────────────┘
```

O desenho mostra estrutura e prioridade. Não precisa ter cores, imagens ou acabamento visual.
